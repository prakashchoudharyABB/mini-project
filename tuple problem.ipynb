```python
# write a program to convert a list into a tuple 
A = [10,20,30,40,50]
B = tuple(A)
print("list:", A)
print("tuple:", B)

```

```python
# write a program to replace last value of tuple in a list to 100
number = [(10,20,30), (40,50,60),(70,80,90)]
A=[]
for tup in number:
    new_tup = tup[:-1] + (100,)
    A.append(new_tup)
print("oiginal list:" ,number)
print("updated list:" ,A)
    
```

```python
#python program to find the maximum and minimum element in a tuple 
A=(25,10,45,30,60)
maximum = max(A)
minimum = min(A)
print("tuple:", A)
print("maximum number:", maximum)
print("minimum number:", minimum)
```
