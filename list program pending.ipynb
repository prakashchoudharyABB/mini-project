```python
list1 = []
n = 20

for i in range(n):
    val = int(input("Enter value: "))
    list1.append(val)

print("List entered:", list1)

# a) similar elements and their index
print("Similar elements and index:")
done = []
for i in range(len(list1)):
    if list1[i] not in done:
        idx = []
        for j in range(len(list1)):
            if list1[j] == list1[i]:
                idx.append(j)
        if len(idx) > 1:
            print(list1[i], "found at index", idx)
        done.append(list1[i])

# b) even and odd count
even = 0
odd = 0
for x in list1:
    if x % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even numbers count:", even)
print("Odd numbers count:", odd)

# c) positive and negative count
pos = 0
neg = 0
for x in list1:
    if x > 0:
        pos = pos + 1
    elif x < 0:
        neg = neg + 1

print("Positive numbers count:", pos)
print("Negative numbers count:", neg)
```

```python
list1 = []
n = 10

for i in range(n):
    val = int(input("Enter value: "))
    list1.append(val)

print("Entered list:", list1)

# a) sorted() ascending
list2 = sorted(list1)
print("Ascending order using sorted():", list2)

# b) sort() descending
list1.sort(reverse=True)
print("Descending order using sort():", list1)

# c) length of list
print("Length of list:", len(list1))
```

```python
list1 = []
list2 = []

n1 = int(input("How many elements in list1: "))
for i in range(n1):
    val = int(input("Enter value: "))
    list1.append(val)

n2 = int(input("How many elements in list2: "))
for i in range(n2):
    val = int(input("Enter value: "))
    list2.append(val)

list3 = list1 + list2
print("Merged list:", list3)
```
