```python
# assign 5 element in dictionary named my_dict • this dictionary contains student's information .
#add and delete student information from my dict. Display student information

my_dict = {"Name": "Prakash","Roll No": 101,"Age": 18,"Course": "B.Tech","Branch": "CSE"}
print("Student Information:")
print(my_dict)
#add student information    
my_dict["College"] = "MIT-WPU"
print("\nAfter adding information:")
print(my_dict)

# Delete student information
del my_dict["Age"]
print("\nAfter deleting information:")
print(my_dict)

# Display student information
print("\nFinal Student Information:")
for key, value in my_dict.items():
 print(key, ":", value)
```
